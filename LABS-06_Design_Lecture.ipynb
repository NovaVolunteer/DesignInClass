# Part 0: How a Jupyter Notebook Works

A **Jupyter notebook** (a file ending in `.ipynb`) is a document that mixes **explanations** and **code you can run** in the same place. It is made of boxes called **cells**. There are two kinds:

- **Markdown cells** (like this one) hold formatted text: headings, bullet points, tables, links. They are for people to read.
- **Code cells** hold Python code. When you *run* a code cell, the computer executes the code and shows the result (called the **output**) directly underneath the cell.

### How to run a cell
Click on a cell, then press **Shift + Enter** (or click the ▶ "Run" button). The notebook runs that cell and moves to the next one.

1. **Run cells from top to bottom.** Later cells depend on earlier ones. If cell 5 uses something created in cell 3, you must run cell 3 first. If you get a strange error, the fix is often: *go back to the top and run every cell in order.*
2. **The number in brackets tells you the order things ran.** Next to each code cell you will see something like `[3]`. That means "this was the 3rd cell run." `[ ]` (empty) means it has not been run yet. `[*]` means it is running right now.
3. **The "kernel" is the engine behind the notebook.** It is the Python program that actually runs your code and remembers everything you have created. If things get really confused, use the menu **Kernel → Restart** (or "Restart Kernel and Run All Cells"). That wipes the notebook's memory, so you then re-run from the top.

## 1.3 Objects, attributes, and methods (the most important idea in this lab)

In Python, almost everything is an **object**. An object is a "thing" that holds **data** and also knows how to **do things** with that data.

A helpful analogy is **a car**:
- A car has **attributes**, facts about it: its color, its number of seats, its mileage. You just *look* at an attribute.
- A car has **methods**, actions it can perform: `start()`, `brake()`, `honk()`. You *ask* it to do something.

In Python we use a **dot** (`.`) to reach into an object:

| | How you write it | Parentheses? | Analogy |
|---|---|---|---|
| **Attribute** | `object.attribute` | **No** `()` | `car.color` → "red" |
| **Method** | `object.method()` | **Yes** `()` | `car.honk()` → *beep!* |

The parentheses are the giveaway. **Parentheses mean "do something."** No parentheses means "tell me a fact you already have."

Here's a real example using a string, which is an object:

````python
word = "data science"

# A METHOD: .upper() is an action. It makes an uppercase version. Note the ().
print(word.upper())

# Another METHOD, this time with a PARAMETER inside the parentheses:
# .replace(old, new) swaps one piece of text for another.
print(word.replace("science", "rocks"))
````

### Parameters (also called *arguments*)

Some methods need extra information to do their job. You give it to them **inside the parentheses**. These inputs are called **parameters** (or *arguments*). Above, `.replace("science", "rocks")` needed to know *what* to replace and *what to replace it with*.

Parameters can also be given **by name**, like `n=10`. You'll see this in the lab with `data.head(n=10)`. Naming the parameter makes the code easier to read.

**Why this matters for the lab:** a pandas DataFrame is an object. The lab uses:
- `data.shape` → an **attribute** (no parentheses)
- `data.head()`, `data.info()`, `data.describe()` → **methods** (parentheses!)

A very common mistake is writing `data.info` without the `()`. Python won't crash. Instead, it just shows you a confusing description of the method itself rather than running it. If you see output like `<bound method DataFrame.info of ...>`, you forgot the parentheses.

````python
print(help(word.upper))  # This shows the documentation for the object "info"
````

# Part 2: What Is Exploratory Data Analysis (EDA)?

When a data scientist gets a new dataset, they don't jump straight to building fancy models. First, they **get to know the data**. This is called **Exploratory Data Analysis**, or **EDA**.

Think of it like moving to a new city. Before deciding where to live, you'd walk around: How big is it? What neighborhoods are there? Where are the problem areas? EDA is walking around your data.

During EDA we try to answer:

- **How much data is there?** (How many rows and columns?)
- **What does it look like?** (What's actually in each column?)
- **What type is each column?** (Numbers? Text? Dates?)
- **Is anything missing or weird?** (Blank cells, impossible values, typos)
- **What are the typical values, and how spread out are they?**
- **What patterns or relationships jump out?** (Do two things tend to go up together?)

EDA is also a **design** step. What you notice here shapes which **questions** you can sensibly ask. This is the "design" part of the Design Project:

```
   Get data  ──►  Explore it (EDA)  ──►  Notice patterns  ──►  Ask a question  ──►  Answer it with evidence
                        ▲                                              │
                        └───────────── often you loop back ◄───────────┘
```

# Part 4: Understanding Your Data

Our data is now stored in a **pandas DataFrame** called `data`.

## 4.1 What is a DataFrame?

A **DataFrame** is a table, a lot like a spreadsheet in Excel or Google Sheets:

- Each **row** is one *observation*: one thing we measured. (Here: one car. In the lab: one movie.)
- Each **column** is one *variable*: one kind of measurement. (Here: mpg, weight… In the lab: budget, revenue, score…)
- Each row has a label on the far left called the **index**. By default it's just 0, 1, 2, 3, … Note that **Python starts counting at 0, not 1.**
- Different columns can hold different types of data (numbers in one, text in another).

A DataFrame is an **object** (remember Part 1.3), so it has **attributes** (facts) and **methods** (actions). We'll now learn four of them, the same four the lab uses:

| Tool | Kind | What it tells you |
|---|---|---|
| `.shape` | attribute | How many rows and columns |
| `.head()` | method | The first few rows, so you can *see* the data |
| `.info()` | method | Column names, data types, and missing values |
| `.describe()` | method | Summary statistics for the numeric columns |

## 4.2 `.shape`: how big is the data?

The [`.shape`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.shape.html) **attribute** tells us the dimensions of the table. Because it's an attribute, there are **no parentheses**.

It gives back two numbers in this format: **(number of rows, number of columns)**.

````python
data.shape
````

### How to read this output

The result `(398, 9)` means:
- **398 rows** → the dataset contains 398 cars
- **9 columns** → we have 9 pieces of information about each car

The parentheses around the two numbers mean it's a **tuple**, Python's word for a fixed, ordered pair (or list) of values. You can grab just one of the numbers using **square brackets** and a position number. Remember Python counts from 0:

````python
print("Number of rows (cars):   ", data.shape[0])   # position 0 = first number
print("Number of columns (vars):", data.shape[1])   # position 1 = second number
````

## 4.3 `.head()`: look at the actual data

Knowing there are 398 rows is useful, but we also want to *see* what the data looks like. We don't need all 398 rows, just enough to understand the layout. The [`.head()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.head.html) **method** shows the first *n* rows.

- It's a **method**, so it needs parentheses: `data.head()`
- By default it shows **5** rows.
- You can choose how many with the parameter `n`. For example, `data.head(n=10)` shows 10 rows. (`data.head(10)` works too.)

````python
data.head(n=10)
````

### What to look for when you view the head

Don't just glance at it. Read it like a detective:

- **What does one row represent?** (Here: one car model.)
- **What kinds of values does each column hold?** `mpg` has decimals. `cylinders` is a small whole number. `origin` and `name` are text.
- **Are the units obvious?** Is `weight` in pounds or kilograms? (Pounds. A 3,500 lb car is normal; a 3,500 kg car would be a truck.) In the movie data, think about whether `budget_x` and `revenue` are in dollars, and how large the numbers are.
- **Anything surprising?** Blank cells shown as `NaN`, strange codes, or columns you don't understand.

**Related tools you might find handy:**
- `data.tail()` shows the **last** rows.
- `data.sample(5)` shows 5 **random** rows. This is often better than `head()` because the top of a file is sometimes unusual (e.g., sorted by date).
- `data.columns` (an attribute!) lists just the column names.

````python
data.columns
````

## 4.4 `.info()`: data types and missing values

Next we want to know **what type of data is in each column** and **whether any values are missing**. The [`.info()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.info.html) **method** gives a compact summary of the whole DataFrame.

Because it's a method, you call it with parentheses: `data.info()`. It **prints** its report directly underneath the cell.

````python
data.info()
````

### How to read `.info()` output, line by line

| Part of the output | What it means |
|---|---|
| `<class 'pandas.core.frame.DataFrame'>` | Confirms that `data` is a DataFrame |
| `RangeIndex: 398 entries, 0 to 397` | There are 398 rows, labeled 0 through 397 |
| `Data columns (total 9 columns)` | There are 9 columns |
| `#` | Column position (starting from 0) |
| `Column` | The column's name |
| `Non-Null Count` | How many rows have a real value (are **not** missing) in that column |
| `Dtype` | The **data type** stored in that column |
| `memory usage` | How much computer memory the table uses (you can ignore this for now) |

### Understanding the `Dtype` column

| Dtype | Meaning | Examples here |
|---|---|---|
| `int64` | **Integer**: whole numbers | `cylinders`, `weight`, `model_year` |
| `float64` | **Floating point**: numbers that can have decimals | `mpg`, `displacement`, `horsepower`, `acceleration` |
| `object` | Usually **text** (strings). pandas uses `object` for anything that isn't a plain number: strings, lists, mixed values | `origin`, `name` |

(The `64` just refers to how much memory each value uses. You don't need to worry about it.)

> **Watch out:** a column that *should* be numbers sometimes shows up as `object`. This usually happens because a few cells contain text, like `"$1,000"` or `"unknown"`. When you see a numeric-sounding column listed as `object`, that's a clue the data needs cleaning.

### Understanding "non-null" and missing values

**Null** means **missing**: the cell is empty. pandas displays missing values as `NaN` ("Not a Number").

**Non-null** means *not missing*. So "Non-Null Count" is the number of rows that actually have a value.

**How to find missing values:** compare each column's non-null count to the total number of rows (398).
- Most columns have `398 non-null` → nothing is missing.
- `horsepower` has `392 non-null` → **398 − 392 = 6 cars are missing their horsepower.**

There's also a direct way to count missing values in every column:

````python
# .isna() marks each missing cell as True, and .sum() counts the Trues in each column
data.isna().sum()
````

Why does missing data matter? Many calculations and plots silently **skip** missing values, so your results are based on fewer rows than you think. If a lot of data is missing, or it's missing for a *reason* (for example, only old movies are missing revenue), your conclusions could be biased. During EDA you're not fixing this yet, just **noticing** it.

## 4.5 `.describe()`: summary statistics

Now we want to **summarize** the numeric columns: what's typical, how spread out the values are, and whether there are extreme values. The [`.describe()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html) **method** calculates a set of **summary statistics** for every numeric column at once.

It's a method, so: `data.describe()`. Unlike `.info()`, which prints a report, `.describe()` **returns a new DataFrame** of statistics, so it shows up as a nice table.

````python
data.describe()
````

### What each statistic means (no stats background needed)

Let's use the **`mpg`** column as our example. Read *down* the `mpg` column of the table above.

| Row | Meaning in plain English | `mpg` example |
|---|---|---|
| **count** | How many non-missing values were used | 398 cars |
| **mean** | The **average**: add all values, divide by the count | The average car gets about 23.5 mpg |
| **std** | **Standard deviation**: roughly, *how far a typical value is from the mean*. Small std = values bunched together. Large std = values spread out | About 7.8. Many cars are within ~8 mpg of the average |
| **min** | The smallest value | The worst car gets 9 mpg |
| **25%** | The **1st quartile**: 25% of cars are *at or below* this value | A quarter of cars get ≤ 17.5 mpg |
| **50%** | The **median**, the middle value: half are below, half above | The middle car gets 23 mpg |
| **75%** | The **3rd quartile**: 75% of cars are at or below this value | Three quarters of cars get ≤ 29 mpg |
| **max** | The largest value | The best car gets 46.6 mpg |

**Quartiles** split the sorted data into four equal-sized groups, like lining all 398 cars up from worst to best mpg and cutting the line at the 1/4, 1/2, and 3/4 marks.

### Three things to look for in `.describe()`

1. **Compare the mean to the median (50%).** If the mean is much **larger** than the median, a few very large values are pulling the average up. Statisticians call this a *right-skewed* distribution. Money columns (like movie **revenue** and **budget**) are almost always like this: a few blockbusters earn billions, while most movies earn far less.
2. **Look at min and max for impossible or suspicious values.** A movie with a budget of `0`? A score of `0`? These are often placeholders for "unknown" rather than real values. Note them.
3. **Check the count row.** If a column's count is lower than the number of rows, that column has missing values. Here, `horsepower` shows 392, which matches what `.info()` told us.

### Reading scientific notation

With large numbers (like movie revenue), pandas may display values like `2.5e+08`. This is **scientific notation**: `e+08` means "× 10⁸", i.e., move the decimal point 8 places to the right. So `2.5e+08` = 250,000,000 (250 million), and `1.2e+09` = 1,200,000,000 (1.2 billion).

### Bonus: summarizing text columns

By default `.describe()` only summarizes numeric columns. To summarize text (`object`) columns, add a parameter:

````python
data.describe(include="object")
````

For text columns you get **count**, **unique** (how many different values), **top** (the most common value), and **freq** (how many times that top value appears). Here, `usa` is the most common origin, appearing 249 times.

# Part 5: Visualizing the Data

Tables of numbers are useful, but humans are much better at spotting patterns in **pictures**. A good chart can reveal in one second what a table hides. In this part we'll make the same three chart types the lab uses:

| Chart | Best for | Lab example | Our example |
|---|---|---|---|
| **Scatterplot** | The **relationship between two numeric** variables | budget vs. revenue | weight vs. mpg |
| **Bar chart** | **Comparing categories** (counts or amounts) | movies per country | cars per origin |
| **Density plot** | The **distribution (shape)** of one numeric variable | movie score | mpg |

## 5.1 The two plotting packages

- **Matplotlib** (`plt`) is the foundational plotting package. It gives you fine control over every piece of the chart.
- **Seaborn** (`sns`) is built *on top of* matplotlib. It makes common statistical charts with less code and nicer defaults. Since seaborn uses matplotlib underneath, you can use `plt` commands (like `plt.title()`) to customize seaborn plots.

## 5.2 The basic recipe for a matplotlib chart

Almost every matplotlib chart in the lab follows the same four steps:

```python
fig, ax = plt.subplots(figsize=(15, 10))   # 1. Create a blank canvas
ax.scatter(x_values, y_values)             # 2. Draw something on it
plt.xlabel("..."); plt.ylabel("...")       # 3. Label it
plt.title("...")
plt.show()                                 # 4. Display it
```

**What is `fig, ax`?** `plt.subplots()` creates two objects at once:
- `fig` (**figure**) is the whole image, the entire canvas or "picture frame."
- `ax` (**axes**) is the actual plotting area inside it: the region with the x-axis and y-axis where the data gets drawn. (Confusingly, "axes" here means *one plot*, not the two axis lines.)

We write `fig, ax = ...` because the function gives back two things, and this stores them in two variables at once.

**`figsize=(15, 10)`** sets the size of the figure as (width, height) in inches. Bigger numbers mean a bigger chart.

## 5.3 Scatterplot

A [scatterplot](https://matplotlib.org/stable/gallery/shapes_and_collections/scatter.html) draws **one dot per row**. Each dot's horizontal position (x) comes from one variable, and its vertical position (y) comes from another. It's the go-to chart for asking **"How does one number relate to another?"**

To pull a single column out of a DataFrame, use **square brackets with the column name in quotes**: `data['weight']`. A single column is called a **Series**.

Let's ask: *How does a car's weight relate to its fuel economy?*

````python
fig, ax = plt.subplots(figsize=(15, 10))          # 1. blank canvas, 15 x 10 inches
ax.scatter(data['weight'], data['mpg'])           # 2. x = weight, y = mpg (one dot per car)

plt.xlabel("Weight (pounds)")                     # 3. label the x-axis...
plt.ylabel("Fuel economy (miles per gallon)")     #    ...and the y-axis
plt.title("Heavier cars get worse fuel economy")  #    ...and give it a title
plt.show()                                        # 4. display the chart
````

### How to read a scatterplot

When you look at a scatterplot, describe these four things:

1. **Direction:** As x goes up, does y tend to go **up** (*positive* relationship), go **down** (*negative* relationship), or show no pattern?
   → Here, as weight goes **up**, mpg goes **down**. This is a **negative** relationship.
2. **Form:** Do the points follow roughly a **straight line**, or a **curve**?
   → Here it's a gentle curve that flattens out for heavy cars.
3. **Strength:** Are the points tightly packed around the pattern (**strong**), or loosely scattered (**weak**)?
   → Fairly tight, so this is a **strong** relationship.
4. **Outliers and clusters:** Are there points far from the rest? Groups of points?
   → A few light cars have unusually high mpg.

**A good title states the takeaway**, not just the variable names. "Heavier cars get worse fuel economy" tells the reader what to see. "Weight vs. MPG" makes them work it out.

### Adding a third variable with color

The lab mentions that you can add features like **color** to show more variables. Each origin gets its own color below. We do this by drawing one group at a time, using a **for loop**. A loop is Python's way of saying "repeat this code once for each item in a list."

````python
fig, ax = plt.subplots(figsize=(15, 10))

# Repeat the indented code once for each origin, with a matching color
for origin, color in [("usa", "tab:blue"), ("europe", "tab:orange"), ("japan", "tab:green")]:
    subset = data[data['origin'] == origin]     # keep only the rows for this origin
    ax.scatter(subset['weight'], subset['mpg'], color=color, label=origin, alpha=0.7)

plt.xlabel("Weight (pounds)")
plt.ylabel("Fuel economy (miles per gallon)")
plt.title("Weight vs. fuel economy, colored by where the car was made")
plt.legend(title="Origin")   # the legend is the key that explains the colors
plt.show()
````

Two new pieces here:
- `data[data['origin'] == origin]` is **filtering**. `data['origin'] == "usa"` checks every row and gives `True` where the origin is usa. Putting that inside `data[ ... ]` keeps only the `True` rows.
- `alpha=0.7` makes dots slightly see-through (0 = invisible, 1 = solid), so overlapping dots are easier to see.

The color reveals something new: **US-made cars were mostly heavier and less fuel efficient**, while Japanese and European cars clustered in the light, high-mpg corner. Adding a third variable gave us a *new insight*, and maybe a new question to ask.

## 5.5 Density plot

### First, what is a "distribution"?

The **distribution** of a variable describes **which values occur and how often**. Are most cars around 20 mpg? Are there two separate groups? Is there a long tail of very efficient cars?

The classic way to see a distribution is a **histogram**. It chops the number line into "bins" (like 10–15 mpg, 15–20 mpg, …) and draws a bar showing how many cars fall in each bin:

````python
plt.figure(figsize=(15, 6))
plt.hist(data['mpg'], bins=20, color="lightgray", edgecolor="black")
plt.xlabel("mpg")
plt.ylabel("number of cars")
plt.title("Histogram of fuel economy")
plt.show()
````

### From histogram to density plot

A [density plot](https://seaborn.pydata.org/generated/seaborn.kdeplot.html) shows the same idea as a histogram, but as a **smooth curve** instead of blocky bars. Imagine laying a smooth sheet over the tops of the histogram bars.

- The technique is called a **KDE**, short for **Kernel Density Estimate**. The idea: put a tiny smooth "bump" on top of every single data point, then add all the bumps together. Where many points are close together, the bumps pile up into a **peak**. Where there are few points, the curve stays **low**. (This is what `gaussian_kde` from `scipy`, which we imported, calculates. "Gaussian" just means each bump is bell-shaped.)
- The y-axis is labeled **Density**. You don't need to interpret the exact number. **Higher means more common**, and the total area under the curve always equals 1 (100% of the data).
- Density plots make it easy to **compare distributions** and spot their **shape**.

In seaborn, a density plot is one line: `sns.kdeplot(data=data, x="mpg")`. You give it the DataFrame (`data=`) and the column name for the x-axis (`x=`).

### Adding a title

Seaborn draws the plot, but it doesn't add a title for you. Since seaborn is built on matplotlib, we add a title with **`plt.title("...")`**. It goes **after** the plot is drawn and **before** `plt.show()`. (Axis labels work the same way: `plt.xlabel()`, `plt.ylabel()`.)

````python
plt.figure(figsize=(15, 10))           # create a new, blank 15 x 10 figure
sns.kdeplot(data=data, x="mpg")        # draw the smooth density curve of mpg

plt.title("Distribution of fuel economy (mpg)")   # <-- the title goes here, before plt.show()
plt.xlabel("Fuel economy (miles per gallon)")
plt.show()
````

> Note: this cell uses `plt.figure(figsize=...)` rather than `fig, ax = plt.subplots(...)`. Both create a blank canvas. `plt.figure()` is a simpler version that's common with seaborn. Don't let the difference confuse you.

### How to read a density plot

Look for:
1. **Peak(s):** Where is the curve highest? That's the most common range of values. → Most cars are around **15–25 mpg**.
2. **Number of peaks:** One peak or two or more? Two peaks (*bimodal*) often means two different groups are mixed together.
3. **Symmetry / skew:** Is the curve balanced, or does it have a long **tail** on one side? → Here there's a longer tail to the **right**: a smaller number of very efficient cars. That's a **right skew**, and it's why the mean (23.5) is a bit above the median (23).
4. **Range:** Where does the curve start and end? → Roughly 5 to 50 mpg. (The smooth curve can extend slightly beyond the real min and max. That's normal for KDE.)

### Bonus: comparing groups with `hue`

Density plots really shine when comparing groups. The `hue` parameter draws a separate curve for each category:

````python
plt.figure(figsize=(15, 8))
sns.kdeplot(data=data, x="mpg", hue="origin", fill=True, common_norm=False, alpha=0.4)
plt.title("Japanese and European cars tend to be more fuel efficient than US cars")
plt.xlabel("Fuel economy (miles per gallon)")
plt.show()
````

(`fill=True` shades under each curve. `common_norm=False` gives each group's curve its own total area of 1, so small groups aren't squashed flat next to big ones.)

The US curve peaks around 15 to 20 mpg, while Japan's peaks around 30. This is a clear difference between groups, and exactly the kind of thing that could become your **question**.

# Part 6: Correlation and Heatmaps

The last code cell in the lab makes a **correlation matrix heatmap**. Let's build up to it.

## 6.1 What is correlation?

**Correlation** is a single number that measures **how strongly two numeric variables move together in a straight-line pattern**. It's usually written as **r**, and it always falls between **−1 and +1**:

| r value | Meaning | Picture it as… |
|---|---|---|
| **+1** | Perfect positive relationship: when one goes up, the other always goes up | Dots on a perfect upward line |
| around **+0.7** | Strong positive | Dots in a tight upward cloud |
| around **+0.3** | Weak positive | Dots in a loose, slightly upward cloud |
| **0** | **No straight-line relationship** | A shapeless blob |
| around **−0.7** | Strong negative: when one goes up, the other tends to go down | Dots in a tight downward cloud |
| **−1** | Perfect negative relationship | Dots on a perfect downward line |

- The **sign** (+ or −) tells you the **direction**.
- The **size** (how close to 1 or −1, ignoring the sign) tells you the **strength**.

Rough rule of thumb: |r| below 0.3 is weak, 0.3 to 0.7 is moderate, and above 0.7 is strong.

Seeing helps. The code below makes **fake data** with different correlations so you can see what each looks like. (You don't need to understand this code, just the pictures.)

````python
rng = np.random.default_rng(1)          # a random number generator (seeded so results repeat)
targets = [0.95, 0.6, 0.0, -0.6, -0.95]

fig, axes = plt.subplots(1, 5, figsize=(20, 4))   # 1 row of 5 small plots
for ax, r in zip(axes, targets):
    x = rng.normal(size=300)
    y = r * x + np.sqrt(1 - r**2) * rng.normal(size=300)   # make y correlated with x by about r
    ax.scatter(x, y, s=8, alpha=0.6)
    ax.set_title(f"r = {np.corrcoef(x, y)[0, 1]:.2f}")
    ax.set_xticks([]); ax.set_yticks([])
plt.suptitle("What different correlation values look like", y=1.05, fontsize=14)
plt.show()
````

## 6.2 Three important warnings about correlation

1. **Correlation is not causation.** Two things moving together doesn't mean one *causes* the other. Ice cream sales and drowning deaths are positively correlated, but ice cream doesn't cause drowning. **Summer** (hot weather) drives both. A hidden third factor like this is called a **confounding variable**. In the lab, a high budget–revenue correlation does *not* prove that spending more *causes* higher revenue. Big franchises, for example, get big budgets *and* big audiences.
2. **Correlation only measures *straight-line* (linear) patterns.** Two variables can have a strong *curved* relationship and still have r near 0. Always look at a scatterplot, not just the number.
3. **Outliers can distort it.** A few extreme points (like a handful of billion-dollar blockbusters) can push r up or down a lot.

## 6.3 Computing correlation with pandas

The `.corr()` **method** computes the correlation between **every pair** of columns you give it, and returns the results as a table called a **correlation matrix**.

To pick several columns at once, use **double square brackets** with a *list* of column names: `data[['mpg', 'weight', 'horsepower']]`. (The outer brackets mean "select from the DataFrame." The inner brackets make a Python **list** of names.)

````python
correlation_matrix = data[['mpg', 'weight', 'horsepower', 'acceleration']].corr()
correlation_matrix
````

### How to read a correlation matrix

- Find the **row** for one variable and the **column** for another. The cell where they meet is their correlation. E.g., row `mpg`, column `weight` → **−0.83**: a strong negative correlation, which matches what the scatterplot showed us.
- The **diagonal** is always **1.00**, because every variable is perfectly correlated with itself. Ignore it.
- The table is **symmetric**: the `mpg`/`weight` cell equals the `weight`/`mpg` cell. So you really only need to read one triangle.

> `.corr()` only works on **numeric** columns. If you include a text column, you'll get an error. Only include numeric columns (int or float), as the lab does with `score`, `budget_x`, and `revenue`.

## 6.4 Visualizing it as a heatmap

A table of numbers is hard to scan. A **heatmap** colors each cell based on its value, so strong relationships jump out visually. This is exactly the structure of the last cell in the lab:

````python
correlation_matrix = data[['mpg', 'weight', 'horsepower', 'acceleration']].corr()

plt.figure(figsize=(15, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix Heatmap')
plt.show()
````

### What each parameter in `sns.heatmap(...)` does

| Parameter | Meaning |
|---|---|
| `correlation_matrix` | The table of numbers to color in |
| `annot=True` | "Annotate": write the number inside each colored square |
| `cmap='coolwarm'` | The **color map**: **blue (cool) = negative**, **red (warm) = positive**, pale/white ≈ 0. The color bar on the right is the key |
| `fmt=".2f"` | Format each number as a decimal with **2** places ("f" = float) |

### Reading this heatmap

- **Deep red** squares (off the diagonal): `weight` and `horsepower` (+0.86). Heavier cars tend to have more powerful engines.
- **Deep blue** squares: `mpg` with `weight` (−0.83) and with `horsepower` (−0.78). Heavier, more powerful cars get worse fuel economy.
- `acceleration` with `horsepower` is blue (−0.69). Remember that acceleration is *seconds to 60 mph*, so a *lower* number is faster. More horsepower means fewer seconds. **Always know your units before interpreting the sign!**

# Part 7: Developing and Answering a Question

This is the heart of the **Design Project**. EDA isn't the goal. It's how you find a **good question**. Now we'll walk through the full process using the car data, so you can repeat it with the movie data.

## 7.1 What makes a good data question?

A good question is:

| Quality | Bad example | Better example |
|---|---|---|
| **Specific**: names the variables | "Are cars good?" | "Do heavier cars get lower mpg?" |
| **Answerable with *this* data**: uses columns you actually have | "Do safer cars sell better?" (no safety or sales columns) | "Did fuel economy improve from 1970 to 1982?" |
| **Measurable**: you can point to a number or chart as the answer | "Is Japan better at cars?" | "Do Japanese cars have higher average mpg than US cars?" |
| **Interesting**: the answer would matter to someone | "How many rows are there?" | "Did the 1970s oil crises push carmakers toward efficiency?" |

**Where do questions come from? Your EDA!** Look back at what surprised you:
- The scatterplot showed a strong weight → mpg pattern.
- The colored scatterplot and the density plot showed that **US cars were less efficient**.
- `.describe()` showed `model_year` runs from 70 to 82, and that decade included two major oil crises (1973 and 1979), when gas prices spiked.

## 7.2 Worked example

### Step 1: State the question (and your guess)

> **Question:** *Did the fuel economy of cars improve between 1970 and 1982, and did US carmakers improve as much as Japanese and European carmakers?*

> **Hypothesis (our prediction before looking):** *Fuel economy improved over the decade because of the oil crises. US cars started far behind and probably improved, but were still behind by 1982.*

Writing a prediction **before** you look keeps you honest. You can't quietly change your story to match whatever you find.

### Step 2: Identify which columns you need
- `mpg` (the outcome we care about)
- `model_year` (time)
- `origin` (the groups we're comparing)

### Step 3: Summarize with numbers using `groupby`

`groupby` is one of the most useful tools in pandas. It splits the data into groups, computes something for each group, and combines the results. In plain English, `data.groupby('model_year')['mpg'].mean()` says: *"Group the cars by model year, then for each year, take the average mpg."*

````python
avg_mpg_by_year = data.groupby('model_year')['mpg'].mean()
avg_mpg_by_year.round(1)
````

Average mpg rose from about **17.7 in 1970** to about **32 in 1982**, nearly doubling. Now let's split by origin as well:

````python
# Group by two things at once: year AND origin. .unstack() spreads origin into columns.
avg_by_year_origin = data.groupby(['model_year', 'origin'])['mpg'].mean().unstack()
avg_by_year_origin.round(1)
````

### Step 4: Visualize it

A **line chart** is the natural choice when the x-axis is **time**:

````python
fig, ax = plt.subplots(figsize=(15, 8))
colors = {"usa": "#232D4B", "europe": "#E57200", "japan": "seagreen"}

for origin in ["usa", "europe", "japan"]:
    ax.plot(avg_by_year_origin.index + 1900,       # turn 70 into 1970, etc.
            avg_by_year_origin[origin],
            marker="o", linewidth=2.5, color=colors[origin], label=origin)

ax.axvline(1973, color="gray", linestyle="--"); ax.text(1973.1, 40, "1973 oil crisis", color="gray")
ax.axvline(1979, color="gray", linestyle="--"); ax.text(1979.1, 40, "1979 oil crisis", color="gray")

plt.xlabel("Model year")
plt.ylabel("Average fuel economy (mpg)")
plt.title("Fuel economy rose for every region from 1970 to 1982; US cars improved but stayed behind")
plt.legend(title="Origin")
plt.show()
````

### Step 5: Answer the question with evidence

> **Answer:** Yes, fuel economy improved substantially. Average mpg across all cars rose from about 18 in 1970 to about 32 in 1982. US cars improved too, climbing steadily almost every year after the 1973 oil crisis (from about 15 mpg in 1973 to about 29 mpg in 1982). But US cars stayed below Japanese and European cars in every single year. Our hypothesis was mostly supported.

Notice how the answer **cites specific numbers** from our analysis and **points to the chart**.

### Step 6: State the limitations (this is what separates good analysis from great analysis)

- **Correlation, not causation:** The timing lines up with the oil crises, but this data can't *prove* the crises caused the change. Government fuel-economy regulations (introduced in the US in 1975) and other factors happened at the same time.
- **The sample may not represent all cars:** We have about 30 cars per year, and they may not reflect what people actually *bought*. A popular gas-guzzler counts the same as a rare compact.
- **Missing data:** 6 cars lacked horsepower. It didn't affect this question (we didn't use horsepower), but it would matter for others.
- **Confounding variable: weight.** Cars also got lighter over this period. Is the mpg gain about *technology* or just *smaller cars*? That would be a great follow-up question.

# Part 8: Troubleshooting Common Errors

Errors are **normal**. Professional data scientists see them all day. When Python hits an error, it prints a **traceback**, a long, scary-looking message. **Read the LAST line first.** It tells you the type of error and usually a hint about what's wrong.

| Error (last line of traceback) | What it means | Common fix |
|---|---|---|
| `FileNotFoundError: No such file or directory: 'imdb_movies.csv'` | Python can't find the file where it's looking | Put the CSV in the same folder as the notebook, or use the full path (see Part 3.3) |
| `NameError: name 'data' is not defined` | You're using a variable that doesn't exist yet. Usually you skipped the cell that creates it, or restarted the kernel | Run the earlier cells (especially the imports and `read_csv`) from the top |
| `NameError: name 'pd' is not defined` (or `plt`, `sns`) | The import cell hasn't been run | Run the import cell first |
| `KeyError: 'Revenue'` | That column name doesn't exist. Column names are **case-sensitive** and must match exactly | Run `data.columns` and copy the exact name (e.g., `'revenue'`, `'budget_x'`) |
| `ModuleNotFoundError: No module named 'seaborn'` | The package isn't installed | Run `%pip install seaborn`, then restart the kernel |
| `SyntaxError` | A typo in the code structure: a missing quote, parenthesis, or comma | Check that every `(` has a `)` and every `"` has a matching `"` |
| Output says `<bound method DataFrame.info of ...>` | Not technically an error. You forgot the parentheses on a method | Write `data.info()` instead of `data.info` |
| `AttributeError: 'DataFrame' object has no attribute 'describ'` | A typo in an attribute or method name | Check the spelling |

Let's see one on purpose so it's less scary. This cell **will** produce an error. Read the last line:

````python
# This will cause a KeyError on purpose: there is no column named 'MPG' (capital letters)
data['MPG']
````

The last line says `KeyError: 'MPG'`. That's pandas telling us "there's no column called `'MPG'`." The real name is lowercase `'mpg'`. Fix the spelling and it works:

````python
data['mpg'].head()
````

**Debugging strategy:**
1. Read the **last line** of the error.
2. Look at the **line the arrow (`---->`) points to**. That's where things broke.
3. Check spelling, capitalization, and parentheses.
4. When in doubt, **Restart the kernel and run all cells from the top.**
5. Still stuck? Copy the error message into a search engine or an AI assistant and ask it to *explain* the error, not just fix it.

# Glossary

| Term | Plain-English definition |
|---|---|
| **Alias** | A short nickname for a package (e.g., `pd` for pandas), set with `import ... as ...` |
| **Attribute** | A fact stored on an object; accessed with a dot and **no** parentheses (`data.shape`) |
| **Bar chart** | A chart of rectangular bars comparing values across categories |
| **Cell** | One box in a notebook; either code or markdown |
| **Confounding variable** | A hidden third factor that affects two variables and makes them look related |
| **Correlation (r)** | A number from −1 to +1 measuring how strongly two numeric variables move together in a straight line |
| **CSV** | Comma-Separated Values; a plain-text table file |
| **DataFrame** | pandas' table object: rows (observations) × columns (variables) |
| **Density plot / KDE** | A smooth curve showing the distribution of a numeric variable |
| **Distribution** | Which values a variable takes and how often |
| **dtype** | The data type of a column (`int64`, `float64`, `object`, …) |
| **EDA** | Exploratory Data Analysis; getting to know a dataset before analyzing it |
| **float** | A number that can have decimals |
| **Function** | A named command that does a job, e.g. `print()`, `pd.read_csv()` |
| **groupby** | Split data into groups, compute something per group, combine the results |
| **Heatmap** | A grid of colored squares where color represents a number |
| **Index** | The row labels of a DataFrame (0, 1, 2, … by default) |
| **int** | A whole number |
| **Kernel** | The engine that runs a notebook's code and remembers its variables |
| **Mean** | The average: sum of values ÷ count |
| **Median** | The middle value when sorted (the 50% row of `.describe()`) |
| **Method** | An action an object can perform; called with a dot **and** parentheses (`data.head()`) |
| **NaN / null** | A missing value |
| **object (dtype)** | pandas' type for text and other non-numeric values |
| **Outlier** | A value far from the rest of the data |
| **Package / library** | A bundle of pre-written code you can import |
| **Parameter / argument** | Extra information you give a function or method inside its parentheses (`n=10`) |
| **Quartile** | Values that split sorted data into four equal parts (25%, 50%, 75%) |
| **Scatterplot** | A chart with one dot per row, showing the relationship between two numeric variables |
| **Series** | A single column of a DataFrame |
| **Skew** | When a distribution has a long tail on one side (right skew = a few very large values) |
| **Standard deviation (std)** | Roughly, how far a typical value is from the mean; a measure of spread |
| **Traceback** | The error message Python prints; read the last line first |
| **Tuple** | A fixed, ordered group of values in parentheses, like `(398, 9)` |
| **Variable** | A name that stores a value, created with `=` |
